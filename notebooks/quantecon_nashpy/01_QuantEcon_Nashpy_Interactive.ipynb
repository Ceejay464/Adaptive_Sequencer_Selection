## Game card · normal form

The row player chooses a row; the column player chooses a column. Each cell shows **(row payoff, column payoff)**. Both choose without seeing the other's current choice.

| Row / Column | Cooperate | Defect |
|---|---:|---:|
| Cooperate | (3, 3) | (0, 5) |
| Defect | (5, 0) | (1, 1) |

The two payoff arrays are

$$
A=\begin{pmatrix}3&0\\5&1\end{pmatrix},\qquad
B=\begin{pmatrix}3&5\\0&1\end{pmatrix}.
$$

A mixed strategy is a probability vector. Write the row strategy as $x=(x_0,x_1)$ and the column strategy as $y=(y_0,y_1)$. Their expected payoffs are

$$
u_1(x,y)=x^{\mathsf T}Ay,\qquad u_2(x,y)=x^{\mathsf T}By.
$$

A Nash equilibrium is a pair $(x^*,y^*)$ at which neither player gains by deviating alone:

$$
(x^*)^{\mathsf T}Ay^*\geq x^{\mathsf T}Ay^*\quad\text{for every }x,
$$

$$
(x^*)^{\mathsf T}By^*\geq(x^*)^{\mathsf T}By\quad\text{for every }y.
$$

**Baseline result:** both defect; the strategy vectors are $(0,1)$ and $(0,1)$; payoffs are $(1,1)$. The solver outputs below verify this statement.

## 1. Represent the game before solving it
Use **A[i,j]** for the row player's payoff and **B[i,j]** for the column player's payoff. In both displayed matrices, i is the row action and j is the column action. A strategy vector is ordered `[probability of action 0, probability of action 1]`.

Read the short function below line by line. The two libraries have different internal axis conventions; the comments mark the required transpose.


```python
def deviation_gain(A, B, x, y):
    """Largest gain either player could obtain by changing their action alone."""
    row_values = A @ y              # Row player's payoff from each pure action.
    column_values = x @ B           # Column player's payoff from each pure action.
    row_gain = row_values.max() - x @ A @ y
    column_gain = column_values.max() - x @ B @ y
    return max(0.0, float(row_gain), float(column_gain))


def solve_matrix(A, B):
    """Run both libraries on the SAME 2-by-2 game and check every candidate."""
    A, B = np.asarray(A, dtype=float), np.asarray(B, dtype=float)
    if A.shape != (2, 2) or B.shape != (2, 2):
        raise ValueError("This introductory explorer expects two 2-by-2 payoff matrices.")
    if not np.isfinite(A).all() or not np.isfinite(B).all():
        raise ValueError("Payoffs must be finite numbers.")

    # Nashpy takes both matrices in the usual [row action, column action] order.
    nash_game = nash.Game(A, B)

    # QuantEcon Player stores its OWN action on axis 0.
    # Therefore player 2 receives B.T, not B. This transpose is essential!
    qe_game = qe.NormalFormGame((qe.Player(A), qe.Player(B.T)))

    rows, notes = [], []
    for name, solver in [("Nashpy", lambda: list(nash_game.support_enumeration())),
                         ("QuantEcon", lambda: list(qe.support_enumeration(qe_game)))]:
        try:
            with warnings.catch_warnings(record=True) as messages:
                warnings.simplefilter("always")
                candidates = solver()
            if messages:
                notes.append(name + ": a solver warning occurred; tied/degenerate games need care.")
            for x, y in candidates:
                x, y = np.asarray(x, float), np.asarray(y, float)
                # Check probabilities as well as the incentive inequalities.
                valid_probability = (np.isfinite(x).all() and np.isfinite(y).all()
                                     and np.min(x) >= -1e-8 and np.min(y) >= -1e-8
                                     and np.isclose(x.sum(), 1) and np.isclose(y.sum(), 1))
                if not valid_probability:
                    notes.append(name + ": discarded an invalid probability vector.")
                    continue
                regret = deviation_gain(A, B, x, y)
                if regret > 1e-7:
                    notes.append(name + ": discarded a candidate that failed the deviation check.")
                    continue
                rows.append({"Tool": name, "P(row action 0)": float(x[0]),
                             "P(column action 0)": float(y[0]),
                             "Row payoff": float(x @ A @ y), "Column payoff": float(x @ B @ y),
                             "Deviation gain": regret})
        except (ValueError, np.linalg.LinAlgError, RuntimeError) as error:
            notes.append(f"{name}: {type(error).__name__}; inspect ties and try the pure check below.")

    # Exhaustive pure-strategy check: only four outcomes, so easy to inspect by hand.
    pure = []
    for i in range(2):
        for j in range(2):
            if A[i, j] >= A[:, j].max() - 1e-9 and B[i, j] >= B[i, :].max() - 1e-9:
                pure.append((i, j))
    return pd.DataFrame(rows), pure, notes


def explore_matrix(A, B, labels=("Action 0", "Action 1")):
    """Display the game, computed candidates, and a short interpretation prompt."""
    A, B = np.asarray(A), np.asarray(B)
    cells = [[f"({A[i,j]:g}, {B[i,j]:g})" for j in range(2)] for i in range(2)]
    display(pd.DataFrame(cells, index=["Row: " + x for x in labels],
                         columns=["Column: " + x for x in labels]))
    result, pure, notes = solve_matrix(A, B)
    display(result.round(6))
    print("Pure Nash outcomes (row, column):", pure)
    print("A deviation gain near zero verifies the stated candidate, not its behavioral realism.")
    print("Support enumeration need not describe an entire continuum in a degenerate game.")
    for note in notes:
        print(note)
    return result
```

## 2. Predict, then run: the prisoner's dilemma
Before running, compare 5 with 3, then 1 with 0. Which row action is a best response in both columns? Repeat for the column player.


```python
# Monday's prisoner's dilemma. First payoff: row player. Second: column player.
A = np.array([[3, 0], [5, 1]])
B = np.array([[3, 5], [0, 1]])
prisoners_dilemma = explore_matrix(A, B, labels=("Cooperate", "Defect"))

# In this game, Defect is a best response to either opponent action.
# Prediction: both defect, receiving (1, 1), even though (3, 3) is feasible.
assert len(prisoners_dilemma) == 2  # One identical equilibrium returned by each library.
assert np.allclose(prisoners_dilemma["P(row action 0)"], 0)
assert np.allclose(prisoners_dilemma["P(column action 0)"], 0)
```

## 3. A mixed equilibrium: matching pennies
Write the indifference equation for each player. Then compare your answer with both libraries.


| Row / Column | Heads | Tails |
|---|---:|---:|
| Heads | (1, −1) | (−1, 1) |
| Tails | (−1, 1) | (1, −1) |

If the column player chooses Heads with probability $q$, the row player's payoffs from Heads and Tails are $2q-1$ and $1-2q$. Indifference gives

$$
2q-1=1-2q\quad\Longrightarrow\quad q=1/2.
$$

The other player's indifference condition gives the same probability. Both players use $(1/2,1/2)$ and earn expected payoff 0.


```python
# Matching pennies has no pure equilibrium: each player randomizes equally.
A_mp = np.array([[1, -1], [-1, 1]])
B_mp = -A_mp
matching_pennies = explore_matrix(A_mp, B_mp, labels=("Heads", "Tails"))
assert np.allclose(matching_pennies["P(row action 0)"], 0.5)
assert np.allclose(matching_pennies["P(column action 0)"], 0.5)
```

## Saved experiment · change one payoff

This comparison has already been run, so it is visible in the GitHub preview. Start with Coordination. Raising the row player's payoff at (action 1, action 0) from 0 to 3 makes action 1 strictly dominant for the row player. Predict which equilibria disappear, then read the saved tables.

```python
# Baseline coordination: two pure equilibria and one fully mixed equilibrium.
A_coord = np.array([[2, 0], [0, 1]])
B_coord = np.array([[2, 0], [0, 1]])
print("Before: coordination game")
coordination = explore_matrix(A_coord, B_coord)

# Change ONE entry. The new payoff creates a profitable deviation at (0, 0).
A_changed = A_coord.copy()
A_changed[1, 0] = 3
print("After: row payoff A[1,0] raised from 0 to 3")
changed_coordination = explore_matrix(A_changed, B_coord)
assert len(changed_coordination) == 2
assert np.allclose(changed_coordination["P(row action 0)"], 0)
assert np.allclose(changed_coordination["P(column action 0)"], 0)
```

## 4. Change the game
Select a preset, edit any of the eight payoffs, and click **Solve and check**. A changed best response can create, remove or multiply equilibria.

If controls do not render, you can always call `explore_matrix(edited_A, edited_B)` in an ordinary code cell. Saving a notebook preserves output; widgets require rerunning their cell after reopening.


The widgets are for a live Colab/Jupyter session. The saved experiment immediately above provides a complete demonstration when you are only viewing this notebook.

```python
# Each box edits one player's payoff at one outcome. Click Solve after changing it.
presets = {
    "Prisoner's dilemma": ([[3, 0], [5, 1]], [[3, 5], [0, 1]]),
    "Matching pennies": ([[1, -1], [-1, 1]], [[-1, 1], [1, -1]]),
    "Coordination": ([[2, 0], [0, 1]], [[2, 0], [0, 1]]),
}
preset = widgets.Dropdown(options=list(presets), description="Example:")
boxes_A = [widgets.IntText(value=int(v), description=f"A[{i//2},{i%2}]",
                          layout=widgets.Layout(width="165px")) for i, v in enumerate(np.ravel(presets[preset.value][0]))]
boxes_B = [widgets.IntText(value=int(v), description=f"B[{i//2},{i%2}]",
                          layout=widgets.Layout(width="165px")) for i, v in enumerate(np.ravel(presets[preset.value][1]))]
solve_button = widgets.Button(description="Solve and check", button_style="primary")
matrix_output = widgets.Output()


def load_preset(change):
    new_A, new_B = presets[change["new"]]
    for box, value in zip(boxes_A, np.ravel(new_A)):
        box.value = int(value)
    for box, value in zip(boxes_B, np.ravel(new_B)):
        box.value = int(value)


def run_matrix(_=None):
    edited_A = np.array([box.value for box in boxes_A]).reshape(2, 2)
    edited_B = np.array([box.value for box in boxes_B]).reshape(2, 2)
    with matrix_output:
        clear_output(wait=True)
        explore_matrix(edited_A, edited_B)


preset.observe(load_preset, names="value")
solve_button.on_click(run_matrix)
payoff_rows = [widgets.HBox([boxes_A[i], boxes_B[i]]) for i in range(4)]
display(widgets.VBox([preset, widgets.HTML("<b>A = row payoff; B = column payoff</b>")]
                     + payoff_rows + [solve_button, matrix_output]))
run_matrix()
```
